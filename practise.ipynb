```python
#using for loop
a = input("enter your numbers: ")
total = 0

for i in a:
    total += i

print(total)
```

```python
#using while loop
numbers = [1, 2, 1]
total = 0
i = 0

while i < len(numbers):
    total += numbers[i]
    i += 1

print(total)  # 4
```

```python

#find the number of vowels in a string
a = "ITvedant"
vowels = "aeiouAEIOU"
count = 0

for i in a:
    if i in vowels:
        count += 1

print("Number of vowels:", count)  # 3
```
